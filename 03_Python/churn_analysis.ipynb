```python
outlier_columns = [
    "age",
    "household_size",
    "active_products",
    "app_logins_frequency",
    "international_transactions",
    "failed_transactions",
    "tx_count",
    "avg_tx_value",
    "total_tx_volume",
    "support_tickets_count",
    "monthly_transaction_count",
    "average_transaction_value",
    "total_transaction_volume",
    "avg_daily_transactions",
    "customer_tenure",
    "customer_lifetime_value"
]

outlier_summary = []

for col in outlier_columns:
    Q1 = customer_df[col].quantile(0.25)
    Q3 = customer_df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = customer_df[
        (customer_df[col] < lower_bound) |
        (customer_df[col] > upper_bound)
    ]
    
    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": round(
            len(outliers) / len(customer_df) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    "Outlier_Percentage",
    ascending=False
)
```

### Plan 3 — Retention KPIs & Customer Segmentation

```python
# ==========================================
# PLAN 3 - STEP 3.1
# Customer Retention KPIs
# ==========================================

total_customers = customer_clean["customer_id"].nunique()

churned_customers_count = (
    customer_clean["churned"] == 1
).sum()

retained_customers_count = (
    customer_clean["churned"] == 0
).sum()

churn_rate = (
    churned_customers_count / total_customers
) * 100

retention_rate = (
    retained_customers_count / total_customers
) * 100


print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers_count)
print("Retained Customers:", retained_customers_count)
print("Churn Rate:", round(churn_rate, 2), "%")
print("Retention Rate:", round(retention_rate, 2), "%")
```

```python
# ==========================================
# PLAN 3 - STEP 3.2
# Transaction Frequency & Customer Value KPIs
# ==========================================

total_transactions = len(transaction_df)

avg_transactions_per_customer = (
    total_transactions / total_customers
)

transactions_per_customer = (
    transaction_df
    .groupby("customer_id")
    .size()
)

repeat_customers = (
    transactions_per_customer >= 2
).sum()

repeat_customer_rate = (
    repeat_customers / total_customers
) * 100

total_spend = transaction_df["amount"].sum()

avg_customer_spend = (
    total_spend / total_customers
)

avg_transaction_value = (
    transaction_df["amount"].mean()
)


print("Total Transactions:", total_transactions)
print(
    "Average Transactions per Customer:",
    round(avg_transactions_per_customer, 2)
)
print("Repeat Customers:", repeat_customers)
print(
    "Repeat Customer Rate:",
    round(repeat_customer_rate, 2),
    "%"
)
print("Total Customer Spend:", round(total_spend, 2))
print(
    "Average Customer Spend:",
    round(avg_customer_spend, 2)
)
print(
    "Average Transaction Value:",
    round(avg_transaction_value, 2)
)
```

```python
# Add transaction count to customer-level table

customer_clean = customer_clean.merge(
    transactions_per_customer.rename("transaction_count"),
    on="customer_id",
    how="left"
)

customer_clean.head()
```

```python
# ==========================================
# PLAN 3 - STEP 3.3
# Customer Tenure & Transaction Gap
# ==========================================

# Customer tenure
customer_clean["customer_tenure_days"] = (
    customer_clean["last_tx"] - customer_clean["first_tx"]
).dt.days


# Average days between transactions
customer_clean["avg_days_between_transactions"] = (
    customer_clean["customer_tenure_days"] /
    (customer_clean["transaction_count"] - 1)
)


# Transactions per 30 days
customer_clean["transactions_per_30_days"] = (
    customer_clean["transaction_count"] /
    (customer_clean["customer_tenure_days"] / 30)
)


# Check results
customer_clean[
    [
        "customer_id",
        "first_tx",
        "last_tx",
        "transaction_count",
        "customer_tenure_days",
        "avg_days_between_transactions",
        "transactions_per_30_days",
        "recency_days",
        "recency_band",
        "churned"
    ]
].head(10)
```

```python
# ==========================================
# PLAN 3 - STEP 3.6
# Retention Priority Segmentation
# ==========================================

def retention_priority(row):

    status = row["recency_band"]
    frequency = row["frequency_segment"]

    # Active customers
    if status == "Active (0-30 days)":
        return "Healthy / Maintain"

    # At Risk customers
    elif status == "At Risk (31-60 days)":
        if frequency == "High Frequency":
            return "High Priority"
        elif frequency == "Regular Frequency":
            return "Medium Priority"
        else:
            return "Re-engagement"

    # High Risk customers
    elif status == "High Risk (61-90 days)":
        return "Urgent Re-engagement"

    # Inactive customers
    elif status == "Inactive (90+ days)":
        return "Inactive / Win-back"


customer_clean["retention_priority"] = (
    customer_clean.apply(
        retention_priority,
        axis=1
    )
)


# Summary
retention_priority_summary = (
    customer_clean["retention_priority"]
    .value_counts()
    .reset_index()
)

retention_priority_summary.columns = [
    "retention_priority",
    "customer_count"
]

retention_priority_summary["percentage"] = (
    retention_priority_summary["customer_count"]
    / total_customers
) * 100

retention_priority_summary
```

```python
# ==========================================
# PLAN 3 - STEP 3.7
# Customer Value Segmentation
# ==========================================

def value_segment(total_spend):
    if total_spend <= 2.015845e7:
        return "Low Value"
    elif total_spend <= 4.808115e7:
        return "Regular Value"
    elif total_spend <= 1.204450e8:
        return "High Value"
    else:
        return "Very High Value"


customer_clean["value_segment"] = (
    customer_clean["total_spend"]
    .apply(value_segment)
)


# Check distribution
value_summary = (
    customer_clean["value_segment"]
    .value_counts()
    .reset_index()
)

value_summary.columns = [
    "value_segment",
    "customer_count"
]

value_summary["percentage"] = (
    value_summary["customer_count"]
    / total_customers
) * 100

value_summary
```

```python
# ==========================================
# PLAN 3 - STEP 3.9
# RFM-style Customer Combination
# ==========================================

rfm_summary = (
    customer_clean
    .groupby(
        [
            "recency_band",
            "frequency_segment",
            "value_segment"
        ]
    )
    .agg(
        customer_count=("customer_id", "nunique"),
        total_spend=("total_spend", "sum"),
        avg_spend=("total_spend", "mean")
    )
    .reset_index()
    .sort_values(
        "customer_count",
        ascending=False
    )
)

rfm_summary.head(20)
```

```python
# ==========================================
# PLAN 3 - STEP 3.10
# RFM Segmentation vs Historical Churn
# ==========================================

rfm_churn_summary = (
    customer_clean
    .groupby(
        [
            "recency_band",
            "frequency_segment",
            "value_segment"
        ]
    )
    .agg(
        customers=("customer_id", "nunique"),
        churned_customers=("churned", "sum")
    )
    .reset_index()
)

rfm_churn_summary["observed_churn_rate"] = (
    rfm_churn_summary["churned_customers"]
    / rfm_churn_summary["customers"]
) * 100

rfm_churn_summary = (
    rfm_churn_summary
    .sort_values(
        "observed_churn_rate",
        ascending=False
    )
)

rfm_churn_summary.head(20)
```

```python
# ==========================================
# PLAN 3 - STEP 3.11
# Final Customer Segmentation
# ==========================================

def customer_segment(row):

    status = row["recency_band"]
    frequency = row["frequency_segment"]
    value = row["value_segment"]

    # 1. Champions
    if (
        status == "Active (0-30 days)"
        and frequency in ["High Frequency", "Very High Frequency"]
        and value in ["High Value", "Very High Value"]
    ):
        return "Champions"

    # 2. Valuable Active Customers
    elif (
        status == "Active (0-30 days)"
        and value in ["High Value", "Very High Value"]
    ):
        return "Valuable Active"

    # 3. High-Value At Risk
    elif (
        status == "At Risk (31-60 days)"
        and value in ["High Value", "Very High Value"]
    ):
        return "At-Risk Valuable"

    # 4. High-Value High Risk
    elif (
        status == "High Risk (61-90 days)"
        and value in ["High Value", "Very High Value"]
    ):
        return "High-Risk Valuable"

    # 5. High-Value Win-back
    elif (
        status == "Inactive (90+ days)"
        and value in ["High Value", "Very High Value"]
    ):
        return "High-Value Win-back"

    # 6. Standard Re-engagement
    elif status in [
        "At Risk (31-60 days)",
        "High Risk (61-90 days)",
        "Inactive (90+ days)"
    ]:
        return "Standard Re-engagement"

    # 7. Other Active Customers
    else:
        return "Regular Active"


customer_clean["customer_segment"] = (
    customer_clean.apply(
        customer_segment,
        axis=1
    )
)


# Segment summary
segment_summary = (
    customer_clean["customer_segment"]
    .value_counts()
    .reset_index()
)

segment_summary.columns = [
    "customer_segment",
    "customer_count"
]

segment_summary["percentage"] = (
    segment_summary["customer_count"]
    / total_customers
) * 100

segment_summary
```
